```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_google_genai import ChatGoogleGenerativeAI
```

### Analyze the user input Language

## Step1: Prompt

```python
prompt_template=ChatPromptTemplate.from_messages([
    ('human',"You're a helpful assistant,Detect the Language"),
    ('ai','Expected Output:{language}')
])
```

## Step2: LLM

```python
llm=ChatGoogleGenerativeAI(
    model="gemini-3.5-flash",   #gemini-3.5-flash
    api_key='AQ.Ab8RN6KIBte9ut-c2RzEwqnHlTkFivSmQlRdvGIcbEsbFGhUtQ',
    temperature=0.2
)
```

## Step3: OutputParser

```python
parser=StrOutputParser()
```

```python
language_chain=prompt_template | llm | parser
response=language_chain.invoke('Machine Learning chaala easy subject voo chaala brief ayindi isme ds')
```

```python
print(response)
```
