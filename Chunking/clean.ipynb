### convert to lowercase

```python
data=data.lower()
data
```

### Replace the spaces

```python
import re

data=re.sub("\s{2,}",'',data)
data
```

```python
import contractions

data=contractions.fix(data)
data
```

### removing punctuation

```python
import string 

punc=string.punctuation
punc
data=re.sub(f'[{punc}]','',data)
data
```

### Remove Special characters

```python
data=re.sub('\W+','',data)
data
```

### Remove emoji and replacing

```python
import emoji

# emoji.demojize(data)
```

```python
data=emoji.replace_emoji(data)
```

```python
data=re.sub('[^0-9a-zA-Z\s]','',data)
data
```

```python
pip install textblob
```

```python
from textblob import TextBlob

txt=TextBlob(data).correct()
data=str(txt)
```

```python
# txt=TextBlob(data).correct().raw_sentences
# print(' '.join(txt))
```

```python
import spacy

nlp=spacy.load('en_core_web_sm')
```

```python
tokens=nlp(data)
updated_tokens=[token.lemma_ for token in tokens if not token.is_stop]
data=' '.join(updated_tokens)
```

### chunking

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    separators=['\n\n','\n']

)

text_splitter.split_text(data)
```
